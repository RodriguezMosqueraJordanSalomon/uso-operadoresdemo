## Refuerzo 2 · Exponentes y raíces (Semana 2)

**Reglas de exponentes:** $a^m \cdot a^n = a^{m+n}$, $\;a^m \div a^n = a^{m-n}$, $\;(a^m)^n = a^{mn}$, $\;a^0 = 1$, $\;a^{-n} = \dfrac{1}{a^n}$.

**Raíces:** la raíz es la inversa de la potencia. $\sqrt{a\cdot b} = \sqrt{a}\cdot\sqrt{b}$ y $\sqrt{a/b} = \sqrt{a}/\sqrt{b}$, pero **$\sqrt{a+b} \neq \sqrt{a}+\sqrt{b}$**. También $\sqrt[n]{a} = a^{1/n}$.

**Radicales:** solo se suman radicales semejantes, por ejemplo $\sqrt{50}+\sqrt{8} = 5\sqrt{2}+2\sqrt{2} = 7\sqrt{2}$. Para racionalizar: $\dfrac{1}{\sqrt{a}} = \dfrac{\sqrt{a}}{a}$.

````python
print(2**3 * 2**4, 2**(3 + 4))
print(2**6 / 2**2, 2**(6 - 2))
print((2**3)**2, 2**(3 * 2))
print(2**-3, 5**0)
print(math.sqrt(49), 8 ** (1/3))
# La raíz no se distribuye sobre la suma:
print(math.sqrt(9 + 16), math.sqrt(9) + math.sqrt(16))
````

## Refuerzo 3 · Razones, proporciones, porcentajes y regla de tres (Semana 3)

**Razón:** compara dos cantidades, $\dfrac{a}{b}$. **Proporción:** igualdad de dos razones, $\dfrac{a}{b}=\dfrac{c}{d}$, que se cumple si $a\cdot d = b\cdot c$ (producto de extremos igual al de medios).

**Porcentaje:** parte $=\dfrac{\text{porcentaje}}{100}\times \text{total}$. Un aumento del 20 % equivale a multiplicar por 1,20 y un descuento del 15 % a multiplicar por 0,85.

**Regla de tres:**
- **Directa:** si una cantidad aumenta, la otra también. $x = \dfrac{b \cdot c}{a}$.
- **Inversa:** si una aumenta, la otra disminuye. $x = \dfrac{a \cdot b}{c}$.
- **Compuesta:** se combinan varias magnitudes, multiplicando una razón por cada una (directa o inversa).

````python
# Razón y proporción: sede A (45 Mbps, 90 usuarios) y sede B (30 Mbps, 60 usuarios)
print(45 / 90, 30 / 60)
print("¿Proporción?", 45 * 60 == 30 * 90)

# Porcentajes: enlace de 200 Mbps con 150 Mbps de tráfico
print("Utilización (%):", 150 / 200 * 100)
print("Con 20 % más de tráfico (Mbps):", 150 * 1.20)

# Regla de tres directa: 10 MB tardan 4 s, ¿25 MB?
print("Directa (s):", 25 * 4 / 10)
# Inversa: 3 analistas tardan 12 h, ¿6 analistas?
print("Inversa (h):", 3 * 12 / 6)
# Compuesta: 3 analistas, 1200 registros, 4 h. ¿6 analistas, 1800 registros?
print("Compuesta (h):", 4 * (3 / 6) * (1800 / 1200))
````
